## Working with lists

```python
# declare an empty string
bob = ''

# declare a list
beatles = []
```

### What kind of variable is this?

```python
print(type(bob))
print(type(beatles))
```

### Populate and print a list

```python
beatles = ["John", "Paul", "George", "Ringo", "Stuart"]

print(beatles)
```

### Calculate and print the length of a list

```python
# calculate the length of a string
print(len(bob))

# calculate and print the number of beatles (i.e., the length of a list)
print(f"There are {len(beatles)} in the Beatles")
```

### Print an item in a list 
   
   _see PowerPoint slides_

```python
# remind ourselves of our list
print(beatles)

# print the first beatle
print(f"The first Beatle: {beatles[0]}")

# print the last Beatle (many ways to do this)
print(f"the last Beatle: {beatles[4]}")
print(f"the last Beatle: {beatles[-1]}")
print(f"the last Beatle: {beatles[len(beatles)-1]}")
print()
print('Now working on George')
print()
print(beatles[2])
print(beatles[-3])
```

### Remove an item from a list

```python
# Remove the last item in the list (bye, Stuart)
stu = beatles.pop()

# What happened to the Beatles?
print(beatles)
print(stu)
```

### Make a new list

```python
# make a list called 'stones' with the Rolling Stones (Mick, Ron, and Keith)
stones = ["Mick", "Ron", "Keith"]

print(stones)
```

### Add two lists

```python
# add two lists together
supergroup = beatles + stones

# print our new list
print(f"Our supergroup has {len(supergroup)} members: {supergroup}")
```
