## Looping

```python
# destiny's child
destiny  = ['beyonce', 'kelly', 'michelle']
beatles  = ['george', 'paul', 'ringo', 'john']
```

### `append()` adds a single item

```python
# use `append()` to add some old men
destiny.append(beatles)
print(destiny)
```

```python
# find and print Ringo, who is buried in a list within a list
print(destiny[-1][-2].capitalize())
print(destiny[3][-2].capitalize())
```

### looping over a list

```python
# loop over 'destiny'
# for <var> in <list>:
for i in destiny:
    print(i)
```

### checking the `type()`

```python
# what is destiny[1]? string destiny[3]? list
for i in destiny:
    print(f"{type(i)}: {i}")
```

### loop over a list within a list

```python
# loop over the old guys in 'destiny'
for oldman in destiny[3]:
    print(oldman.capitalize())
```

### introducing a test

```python
# what type of variable is this anyway?
for i in destiny:
    if type(i) is list:
        for dude in i:
            print(dude.capitalize())
    else:
        print(i.capitalize())
```

### executing a nested `for` loop

```python
# loop within a loop

#------ counting from 1

# initialize two counters
x = 1 # counting the old guys
y = 1 # counting the divas


for i in destiny:
    if type(i) is list:
        for dude in i:
            print(f"Member number {x} of Destiny's Child is: {dude.capitalize()}")
            x += 1
            # x = x + 1 <- this would also work
    else:
        print(f"Member number {y} of Destiny's Child is: {i.capitalize()}")
        y += 1

#------ counting from 0

print()
print("---- now counting from zero ----")
print()

# initialize two counters
x = 0 # counting the old guys
y = 0 # counting the divas

for i in destiny:
    if type(i) is list:
        for dude in i:
            x += 1
            print(f"Member number {x} of Destiny's Child is: {dude.capitalize()}")
            # x = x + 1 <- this would also work
    else:
        y += 1
        print(f"Member number {y} of Destiny's Child is: {i.capitalize()}")





```

### retrieve the index _and_ the value with `enumerate()`

```python
# define a list of American presidents
presidents = ["George Washington", "John Adams", "Thomas Jefferson", "James Madison", "James Monroe", "John Quincy Adams", "Andrew Jackson", "Martin Van Buren", "William Henry Harrison", "John Tyler", "James Knox Polk", "Zachary Taylor", "Millard Fillmore", "Franklin Pierce", "James Buchanan", "Abraham Lincoln", "Andrew Johnson", "Ulysses S. Grant", "Rutherford Birchard Hayes", "James Abram Garfield", "Chester Alan Arthur", "Grover Cleveland", "Benjamin Harrison", "Grover Cleveland", "William McKinley", "Theodore Roosevelt", "William Howard Taft", "Woodrow Wilson", "Warren Gamaliel Harding", "Calvin Coolidge", "Herbert Clark Hoover", "Franklin Delano Roosevelt", "Harry S. Truman", "Dwight David Eisenhower", "John Fitzgerald Kennedy", "Lyndon Baines Johnson", "Richard Milhous Nixon", "Gerald Rudolph Ford", "James Earl Carter Jr.", "Ronald Wilson Reagan", "George Herbert Walker Bush", "William Jefferson Clinton", "George Walker Bush", "Barack Hussein Obama", "Donald John Trump", "Joe Biden", "Donald John Trump"]

# lots of flexibility with enumerate()
for num, name in enumerate(presidents):
    print(f"President number {num+1} is: {name}")
```

### loop over a string

```python
today  = "Wednesday"
vowels = "aeiou"

for letter in today:
    # print the vowels only
    # if letter in vowels:

    # print the consonants only
    if letter not in vowels:
        print(letter)
```

#### So . . . would you rather get $10,000 or a penny that doubled every day for 30 days?

```python
pennies = 1

for i in range(1,31):
    pennies *= 2
    print(f"Day {i}: {pennies}")

print(f"The grand total: {pennies:,d}")
```
