### Printing quotation marks

##### Write two print statements that give the following output:

She said, 'hello world'

She said, "hello world"


```python
# using mixed quotes to print quotes
print("She said, 'hello world'")
print('She said, "hello world"')

# you can escape "special" characters
print('She said, \'hello world\'')
```

### Store and print a variable

```python
# define some strings
first = 'bob'
last  = 'dylan'

# concatenate those strings
full = first + last

# "+" but with math
sum = 9 + 9

# print the strings
print(first)
print(last)
print(full)
print(sum)
print(first + last + "first") # what's the output?
print(first, last, "first") # how is this different?
print(first + " " + last + " " + "is super old")


```

### Formatted strings are the way to go

```python
print(f'{first} {last} is super old')
print(f"{first} {last} is super old")

print(f'{first} {last} knows the answer to 7+6 is {7+6}')
```

### Get the length of a string

```python
# the long (but readable) way
name_length = len(full)
print(name_length)

# a more concise way to do it, functions within functions
# print this in a single print statement: "There are 8 characters in bobdylan"
print(f"There are {len(full)} characters in {first + last}")
print(f"There are {len(full)} characters in {full}")
```

### <span style="color:red">Problem #1:</span> Write a program that asks for the user's name and age, then prints a personalized greeting using string formatting (`.format`).

For *Andy* and *32*, the output would be:

Hello, Andy! You are 32 years old.


```python
# code here
name = input("Enter your name: ")
age  = input("Enter your age: ")

print("Hello {}, You are {} years old.".format(name, age))
print(f"Hello {name}, You are {age} years old.")
```

### <span style="color:red">Problem #2:</span> Ask the user for the number of decimal places, then print the following output. If the user enters 3, the output will be: 

The value of pi is 3.142

```python
import math

num_decimals = int(input("Enter the number of decimal places: "))

print(f"The value of pi is {round(math.pi, num_decimals)}")
```

### <span style="color:red">Problem #4:</span> Tip Calculator

Ask the user for the bill amount (float) and tip percentage (integer). Calculate the tip and total, then print them neatly.

Enter the bill amount: `42.75`

Enter the tip percentage: `18`

Bill: `$42.75`

Tip (18%): `$7.70`

\--------------------

Total: `$50.45`


```python
# prompt user for input (convert both to floats)
bill        = float(input('Enter the bill amount: '))
tip_percent = float(input('Enter the tip percentage: '))

# calculate the tip amount and total bill
tip_amount  =  bill * (tip_percent/100)
total       = bill + tip_amount

print(f"Bill: ${bill}")
print(f"Tip ({tip_percent}%): ${round(tip_amount, 2)}")
print('--------------------')
print(f"Total: ${round(total, 2)}")

```
